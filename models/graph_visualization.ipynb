```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# --- Load image (replace with your file) ---
img = np.array(Image.open("/home/fnu.talha/resvit/realT1ce.png").convert("L"))  # grayscale

# --- Sobel filters implemented manually ---
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]])

def convolve2d(img, kernel):
    h, w = img.shape
    kh, kw = kernel.shape
    pad_h, pad_w = kh//2, kw//2
    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode='reflect')
    
    out = np.zeros_like(img, dtype=float)
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            out[i, j] = np.sum(region * kernel)
    return out

# compute edges
gx = convolve2d(img, sobel_x)
gy = convolve2d(img, sobel_y)
edges = np.sqrt(gx**2 + gy**2)

# --- FFT magnitude spectrum ---
f = np.fft.fft2(img)
fshift = np.fft.fftshift(f)
fft_mag = np.log(np.abs(fshift) + 1)

# --- Visualization ---
plt.figure(figsize=(14,4))

plt.subplot(1,3,1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(edges, cmap="gray")
plt.title("Edges (Sobel – no cv2)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(fft_mag, cmap="gray")
plt.title("FFT Magnitude Spectrum")
plt.axis("off")

plt.show()
```
